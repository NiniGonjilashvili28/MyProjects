```sql

CREATE OR REPLACE TABLE getdata.raw.billionaires_2023 AS
SELECT *
FROM read_files(
  '/Volumes/getdata/raw/raw_files/Billionaires Statistics Dataset.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
 
```

```sql

CREATE OR REPLACE TABLE getdata.calculated.billionaires_clean AS
SELECT
  b.personName                              AS person_name,
  b.age,
  CASE
    WHEN b.age IS NULL          THEN 'უცნობი'
    WHEN b.age < 40              THEN '<40'
    WHEN b.age BETWEEN 40 AND 59 THEN '40-59'
    WHEN b.age BETWEEN 60 AND 79 THEN '60-79'
    ELSE '80+'
  END                                        AS age_bucket,
  b.country,
  b.category                                 AS industry,
  b.selfMade                                 AS is_self_made,
  b.finalWorth                               AS net_worth_million_usd,
  b.gross_tertiary_education_enrollment      AS education_enrollment_pct
FROM getdata.raw.billionaires_2023 AS b
WHERE b.personName IS NOT NULL
  AND b.country    IS NOT NULL;
 
```

```sql
CREATE OR REPLACE TABLE getdata.calculated.era_summary_grouped AS
SELECT
  CASE
    WHEN era IN ('19th_century', '20th_century') THEN 'ისტორიული (19-20 ს.)'
    ELSE 'თანამედროვე (21 ს.)'
  END AS era_group,
  COUNT(*)                                AS people_count,
  ROUND(AVG(years_founding_to_peak), 1)   AS avg_years_to_peak
FROM getdata.calculated.historical_titans
GROUP BY
  CASE
    WHEN era IN ('19th_century', '20th_century') THEN 'ისტორიული (19-20 ს.)'
    ELSE 'თანამედროვე (21 ს.)'
  END;
 
```

```sql

CREATE OR REPLACE TABLE getdata.calculated.success_by_industry AS
WITH industry_counts AS (
  SELECT
    industry,
    COUNT(*)                                          AS total_people,
    SUM(CASE WHEN is_self_made THEN 1 ELSE 0 END)     AS self_made_count,
    AVG(age)                                          AS avg_age
  FROM getdata.calculated.billionaires_clean
  GROUP BY industry
)
SELECT
  industry,
  total_people,
  self_made_count,
  ROUND(self_made_count / total_people * 100, 1)      AS self_made_pct,
  ROUND(avg_age, 1)                                   AS avg_age,
  RANK() OVER (ORDER BY self_made_count / total_people DESC) AS self_made_rank
FROM industry_counts
ORDER BY self_made_pct DESC;
 
```

```sql
CREATE OR REPLACE TABLE getdata.calculated.success_by_country AS
WITH country_counts AS (
  SELECT
    country,
    COUNT(*)                                          AS total_people,
    SUM(CASE WHEN is_self_made THEN 1 ELSE 0 END)     AS self_made_count,
    AVG(education_enrollment_pct)                     AS avg_education_enrollment
  FROM getdata.calculated.billionaires_clean
  GROUP BY country
  HAVING COUNT(*) >= 5   -- გამოვრიცხავთ ქვეყნებს, სადაც ძალიან ცოტა ჩანაწერია (არასაიმედო %)
)
SELECT
  country,
  total_people,
  self_made_count,
  ROUND(self_made_count / total_people * 100, 1)      AS self_made_pct,
  ROUND(avg_education_enrollment, 1)                  AS avg_education_enrollment,
  RANK() OVER (ORDER BY total_people DESC)            AS country_rank
FROM country_counts
ORDER BY total_people DESC;
 
```

```sql

CREATE OR REPLACE TABLE getdata.calculated.leaders_profile AS
SELECT
  bc.person_name,
  bc.age,
  bc.industry,
  bc.country,
  bc.net_worth_million_usd,
  sf.dropped_out,
  sf.first_venture_failed,
  sf.age_at_first_success,
  sf.industry_pivots
FROM getdata.calculated.billionaires_clean AS bc
JOIN getdata.calculated.success_factors    AS sf
  ON bc.person_name = sf.person_name;
```

```sql

CREATE OR REPLACE TABLE getdata.calculated.era_summary AS
SELECT
  era,
  COUNT(*)                                AS people_count,
  ROUND(AVG(years_founding_to_peak), 1)   AS avg_years_to_peak,
  ROUND(AVG(peak_pct_us_gdp), 2)          AS avg_peak_pct_gdp
FROM getdata.calculated.historical_titans
GROUP BY era
ORDER BY avg_years_to_peak;
 
```
